```python
# Step 1: Replicate each row N times for independent sampling
replicate_block = RowMultiplierBlock(
    block_name="expand_samples",
    num_samples=5,  # Each policy row becomes 5 rows
    shuffle=False
)

replicated_dataset = replicate_block.generate(base_dataset)

# Manually add sample_id column (ideally, we'd want sgd-hub to handle this)
# Each group of 5 rows gets indices 0-4
replicated_dataset['sample_id'] = np.tile(np.arange(5), len(base_dataset))

print(f"After replication: {len(replicated_dataset)} rows")
replicated_dataset
```

```python
# Step 2: Sample from pool columns embedded in the dataset
# Each row's pool values are sampled independently
# SamplerBlock handles one column at a time, so we chain multiple calls

# Start with replicated dataset
sampled_dataset = replicated_dataset.copy()

# Define pool mappings (output_col -> pool_col)
pool_mappings = {
    "demographic_group": "demographics_pool",
    "expertise_level": "expertise_pool",
    "region": "geography_pool",
    "lang_style": "language_styles_pool",
    "exploit_stage": "exploit_stages_pool",
    "medium": "task_medium_pool",
    "temporal_context": "temporal_pool",
    "trust_signal": "trust_signals_pool",
}

# Apply SamplerBlock for each pool column
for output_col, pool_col in pool_mappings.items():
    sampler = SamplerBlock(
        block_name=f"sample_{output_col}",
        input_cols=[pool_col],
        output_cols=[output_col],
        num_samples=1,  # Sample 1 value from each row's pool
        random_seed=42,  # Consistent seed across all samplers
    )
    sampled_dataset = sampler.generate(sampled_dataset)
    # Extract single value from list (SamplerBlock returns lists)
    sampled_dataset[output_col] = sampled_dataset[output_col].apply(lambda x: x[0] if x else None)

# Drop pool columns after sampling (ideally we'd want to control this)
pool_cols_to_drop = list(pool_mappings.values())
sampled_dataset = sampled_dataset.drop(columns=pool_cols_to_drop)

print(f"After sampling: {len(sampled_dataset)} rows, {len(sampled_dataset.columns)} columns")
print(f"Columns: {sampled_dataset.columns.tolist()}")
sampled_dataset
```

```python
# Step 3: Build prompts using the sampled dimensions
input_cols = [
    "policy_concept",
    "concept_definition",
    "demographic_group",
    "expertise_level",
    "region",
    "lang_style",
    "exploit_stage",
    "medium",
    "temporal_context",
    "trust_signal"
]

prompt_builder = PromptBuilderBlock(
    block_name="redteam_prompt_builder",
    input_cols=input_cols,
    output_cols="messages",
    prompt_config_path="cb-gen-template.yaml",
    format_as_messages=True
)

result = prompt_builder(sampled_dataset)
print(f"Final dataset: {len(result)} rows")
result
```

```python
import json
import re


def extract_prompt_fields(prompt_list):
    if not prompt_list:
        return {}

    content = prompt_list[0].get('content', '{}')

    # Find the JSON object - extract from first { to last }
    start = content.find('{')
    end = content.rfind('}')

    if start == -1 or end == -1:
        return {'raw_content': content}

    json_str = content[start:end + 1]

    # Fix trailing commas
    json_str = re.sub(r',\s*}', '}', json_str)
    json_str = re.sub(r',\s*]', ']', json_str)

    try:
        return json.loads(json_str)
    except json.JSONDecodeError as e:
        print(f"JSON parse error at position {e.pos}: {e.msg}")
        print(f"Problematic area: ...{json_str[max(0, e.pos - 30):e.pos + 30]}...")
        return {'raw_content': content}


# Parse the JSON content and expand into columns (ideally, we'd be able to do this in sdg-hub but not critical)
parsed = prompts_result['prompt'].apply(extract_prompt_fields).apply(pd.Series)

final_result = pd.concat([
    prompts_result.drop(columns=['prompt', 'messages']),
    parsed
], axis=1)

print(f"Extracted columns: {parsed.columns.tolist()}")
final_result
```
